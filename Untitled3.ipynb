```python

foot_falls = [100, 120, 150, 200, 0, 40, 70, 30, 90]
collections = [1000, 750, 850, 900, 620, 240, 470, 330, 590]
reasons = ["BB1", "sm", "sm2", "BB2", "c19", "sm3", "sm4", "sm5", "BB3"]
```

```python
from matplotlib.cbook import violin_stats

plt.plot(reasons, foot_falls,color='green',linewidth=1,linestyle='--',marker='D')
plt.plot(reasons, collections,color='red',linewidth=1,linestyle='-.', marker='o')
plt.title("theater_footfall")
plt.xlabel("Reasons")
plt.ylabel("Footfall  @ collections")
plt.legend(["footfall","collections"])
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

hours = [1, 2, 3, 4, 5]
rev = [1, 3, 2, 2, 1]
marks = [35, 45, 60, 70, 80]

plt.scatter(hours, marks, color='red',marker='*',s=100)
plt.scatter(rev, marks, color='blue',marker='+',s=100)

plt.title("Study")
plt.xlabel("Hours & rev")
plt.ylabel("Marks")
plt.legend(["hours","rev"])
plt.show()
```
